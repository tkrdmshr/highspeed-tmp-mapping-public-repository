```python
import pandas as pd
from draw_graph import Scatter, Stripplot
import matplotlib.pyplot as plt
from calibration import lifetime_to_temperature_201112, temperature_to_lifetime_201112, lifetime_to_temperature_int_weight_201112, temperature_to_lifetime_int_weight_201112
import numpy as np
import statistics as st
from matplotlib.backends.backend_pdf import PdfPages
```

```python
df_accuracy = pd.read_csv('./tauFAST_accuracy.csv', header = None)
```

```python
df_24deg = df_accuracy.iloc[1:, 1:6]
df_27deg = df_accuracy.iloc[1:, 6:11]
df_30deg = df_accuracy.iloc[1:, 11:16]
df_24deg_ave = df_accuracy.iloc[12, 1:6]
df_27deg_ave = df_accuracy.iloc[12, 6:11]
df_30deg_ave = df_accuracy.iloc[12, 11:16]

fig = plt.figure(figsize = [20, 20])
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['xtick.major.width'] = 2.0
plt.rcParams['ytick.major.width'] = 2.0
plt.rcParams['axes.linewidth'] = 2.0
plt.plot(df_24deg.iloc[0, :], np.array(df_24deg.T.iloc[:, 1:]), marker = 'o', markersize = 10, linewidth = 0, markeredgecolor ='blue', markerfacecolor = 'None')
plt.plot(df_27deg.iloc[0, :], np.array(df_27deg.T.iloc[:, 1:]), marker = 'o', markersize = 10, linewidth = 0, markeredgecolor ='black', markerfacecolor = 'None')
plt.plot(df_30deg.iloc[0, :], np.array(df_30deg.T.iloc[:, 1:]), marker = 'o', markersize = 10, linewidth = 0, markeredgecolor ='red', markerfacecolor = 'None')
plt.plot(df_24deg.iloc[0, :], df_24deg_ave, marker = 'o', markersize = 20, linewidth = 0, color = 'blue')
plt.plot(df_27deg.iloc[0, :], df_27deg_ave, marker = 'o', markersize = 20, linewidth = 0, color = 'black')
plt.plot(df_30deg.iloc[0, :], df_30deg_ave, marker = 'o', markersize = 20, linewidth = 0, color = 'red')

plt.xscale('log')
plt.tick_params(axis='both', direction = "in", labelsize = 30)

plt.xlim(10**3, 10**7)
plt.ylim(4, 12)
plt.show()

#pdfのセーブ
pp = PdfPages('./tauFAST_accuracy.pdf')
pp.savefig(fig)
pp.close()
```
