```python
import pandas as pd
from draw_graph import Scatter, Stripplot
import matplotlib.pyplot as plt
import numpy as np
import statistics as st
from matplotlib.backends.backend_pdf import PdfPages
```

```python
df_24deg = pd.read_csv('./24deg_precision.csv', header = None)
```

```python
df_24deg_extract = df_24deg[df_24deg[0].str.endswith(str(1))].iloc[:, 1:]
photon = df_24deg_extract.iloc[0, :]
tauFAST = df_24deg_extract.iloc[1, :]
tauf = df_24deg_extract.iloc[2, :]

fig = plt.figure(figsize = [20, 20])
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['xtick.major.width'] = 2.0
plt.rcParams['ytick.major.width'] = 2.0
plt.rcParams['axes.linewidth'] = 2.0
plt.plot(photon, tauFAST, marker = 'o', color = 'red', linewidth = 0, markersize = 20)
plt.plot(photon, tauf, marker = 'o', color = 'black', linewidth = 0, markersize = 20)
plt.xscale('log')
plt.tick_params(axis='both', direction = "in", labelsize = 24)

#図の表示
plt.xlim(10**3, 10**7)
plt.ylim(0, 1)
plt.show()

#pdfのセーブ
pp = PdfPages('./revise_tauFAST_tauf_compare_24deg.pdf')
pp.savefig(fig)
pp.close()
```
