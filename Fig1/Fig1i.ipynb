```python
from PIL import Image
import numpy as np
import pandas as pd
from calibration import lifetime_to_temperature_water, lifetime_to_temperature_extract, pict_calibration
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
```

```python
water_lifetime = np.array(Image.open('./Water/Project_32.0_fast_ch01.tif'))/1000
water_tmp = lifetime_to_temperature_water(water_lifetime)
cell_lifetime = np.array(Image.open('./Cell/Project002_10um_fast_1_ch01.tif'))/1000
cell_tmp = lifetime_to_temperature_extract(cell_lifetime)
```

```python
fig = plt.figure(figsize = (20, 20))
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['xtick.major.width'] = 1.0
plt.rcParams['ytick.major.width'] = 1.0
plt.rcParams['axes.linewidth'] = 1.0
plt.rcParams["font.size"] = 20
plt.tick_params(labelbottom=False,
                labelleft=False,
                labelright=False,
                labeltop=False)
plt.tick_params(bottom=False,
                left=False,
                right=False,
                top=False)

plt.imshow(water_tmp-np.mean(water_tmp), vmin = -3, vmax = 3, cmap = 'jet')
#plt.colorbar()
plt.show()

#pdfのセーブ
pp = PdfPages('./fig1i_water.pdf')
pp.savefig(fig)
pp.close()
```
