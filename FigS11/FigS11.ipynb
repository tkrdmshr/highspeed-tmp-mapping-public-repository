```python
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from calibration import lifetime_to_temperature_201112
from scipy.optimize import curve_fit
from scipy import stats
from matplotlib.backends.backend_pdf import PdfPages
from scipy.stats import sem
```

```python
files_path = glob.glob('./5sec/20111*_relaxation_whole_lin22.6_70msec_100mW_5sec/whole_cells/*.csv')
files_path.sort()
```

```python
plt.figure(figsize = (6, 6))
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['xtick.major.width'] = 2.0
plt.rcParams['ytick.major.width'] = 2.0
plt.rcParams['axes.linewidth'] = 2.0
data_set_5sec = pd.DataFrame()

#データの平均化
average = np.zeros((1, 240))
for path in files_path:
    df = pd.read_csv(path, sep = '\t')
    relaxation_int = np.array(df.iloc[1:, 3], dtype = 'float32')
    relaxation = np.array(df.iloc[1:, 1], dtype = 'float32')
    diff = relaxation_int[:-1] - relaxation_int[1:]
    off_time = np.argmax(diff)
    extract = lifetime_to_temperature_201112(relaxation[off_time-100 : off_time+140])
    average += extract
    data_set_5sec = pd.concat([data_set_5sec, pd.DataFrame(extract)], axis = 1)
average_normal = average[0]/len(files_path)
average_normal = average_normal - np.mean(average_normal[:10])
time = [(i-100)*0.07 for i in range(len(average_normal))]
plt.tick_params(axis='both', direction = "out", labelsize = 24)
plt.plot(time, average_normal, color = 'Black')
plt.ylim(-0.2, 2.0)
plt.xlabel('Time (sec)', fontsize = 24)
plt.ylabel('∆T (˚C)', fontsize = 24)
plt.show()

#データの保存
(data_set_5sec - data_set_5sec.iloc[:10, :].mean()).to_csv('5sec_all.csv')
pd.DataFrame(average_normal).to_csv('5sec_deltaT_ave.csv')
```

```python
#SEMの計算
def calc_SEM(df, start):
    df = df - np.mean(df.iloc[:10, :])
    df = df/df.iloc[start, :]*100
    sem_df = df.sem(axis = 1)
    sem_df[start] = 0
    return sem_df

#温度バーを挿入
def make_bar(df, pos, deg, start):
    bar = [100-deg*100/df[start]*i/100 for i in range(100)]
    bar_pos = [pos for i in range(len(bar))]
    return bar_pos, bar
```

```python
#実測データのプロット
fig = plt.figure(figsize = (20, 20))
plt.tick_params(axis='both', direction = "in", labelsize = 24)
time_normal_100msec = [(i)*0.07 for i in range(len(average_normal_100msec[50:100]))]
time_normal_500msec = [(i)*0.07 for i in range(len(average_normal_500msec[50:100]))]
time_normal_1sec = [(i)*0.07 for i in range(len(average_normal_1sec[50:100]))]
time_normal_5sec = [(i)*0.07 for i in range(len(average_normal[100:150]))]
time_normal_10sec = [(i)*0.07 for i in range(len(average_normal_10sec[200:250]))]


#SEMの計算
sem_100msec = calc_SEM(data_set_100msec, 50)
sem_500msec = calc_SEM(data_set_500msec, 50)
sem_1sec = calc_SEM(data_set_1sec, 50)
sem_5sec = calc_SEM(data_set_5sec, 100)
sem_10sec = calc_SEM(data_set_10sec, 200)


#0.2˚Cを示すバーを挿入
bar_100msec_pos, bar_100msec = make_bar(average_normal_100msec, pos = 2.6, deg = 0.2, start = 50)
plt.plot(bar_100msec_pos, bar_100msec, linewidth = 4, color = 'black', linestyle = '-')
bar_500msec_pos, bar_500msec = make_bar(average_normal_500msec, pos = 2.7, deg = 0.2, start = 50)
plt.plot(bar_500msec_pos, bar_500msec, linewidth = 4, color = 'blue', linestyle = '-')
bar_1sec_pos, bar_1sec = make_bar(average_normal_1sec, pos = 2.8, deg = 0.2, start = 50)
plt.plot(bar_1sec_pos, bar_1sec, linewidth = 4, color = 'green', linestyle = '-')
bar_5sec_pos, bar_5sec = make_bar(average_normal, pos = 2.9, deg = 0.2, start = 100)
plt.plot(bar_5sec_pos, bar_5sec, linewidth = 4, color = 'orange', linestyle = '-')
bar_10sec_pos, bar_10sec = make_bar(average_normal_10sec, pos = 3.0, deg = 0.2, start = 200)
plt.plot(bar_10sec_pos, bar_10sec, linewidth = 4, color = 'red', linestyle = '-')


average_normal_100msec = average_normal_100msec/average_normal_100msec[50]
average_normal_500msec = average_normal_500msec/average_normal_500msec[50]
average_normal_1sec = average_normal_1sec/average_normal_1sec[50]
average_normal = average_normal/average_normal[100]
average_normal_10sec = average_normal_10sec/average_normal_10sec[200]

plt.errorbar(time_normal_100msec, average_normal_100msec[50:100]*100/average_normal_100msec[50], yerr = np.ravel(sem_100msec)[50:100], capsize=0, fmt='o', markersize=0, linewidth = 5, ecolor='black')
plt.errorbar(time_normal_500msec, average_normal_500msec[50:100]*100/average_normal_500msec[50], yerr = np.ravel(sem_500msec)[50:100], capsize=0, fmt='o', markersize=0, linewidth = 5, ecolor='blue')
plt.errorbar(time_normal_1sec, average_normal_1sec[50:100]*100/average_normal_1sec[50], yerr = np.ravel(sem_1sec)[50:100], capsize=0, fmt='o', markersize=0, linewidth = 5, ecolor='green')
plt.errorbar(time_normal_5sec, average_normal[100:150]*100/average_normal[100], yerr = np.ravel(sem_5sec)[100:150], capsize=0, fmt='o', markersize=0, linewidth = 5, ecolor='orange')
plt.errorbar(time_normal_10sec, average_normal_10sec[200:250]*100/average_normal_10sec[200], yerr = np.ravel(sem_10sec)[200:250], capsize=0, fmt='o', markersize=0, linewidth = 5, ecolor='red')

plt.plot(time_normal_100msec, average_normal_100msec[50:100]*100/average_normal_100msec[50], color = 'black', markersize = 20, marker = 'o', linewidth = 0)
plt.plot(time_normal_500msec, average_normal_500msec[50:100]*100/average_normal_500msec[50], color = 'blue', markersize = 20, marker = 'o', linewidth = 0)
plt.plot(time_normal_1sec, average_normal_1sec[50:100]*100/average_normal_1sec[50], color = 'green', markersize = 20, marker = 'o', linewidth = 0)
plt.plot(time_normal_5sec, average_normal[100:150]*100/average_normal[100], color = 'orange', markersize = 20, marker = 'o', linewidth = 0)
plt.plot(time_normal_10sec, average_normal_10sec[200:250]*100/average_normal_10sec[200], color = 'red', markersize = 20, marker = 'o', linewidth = 0)

#plt.ylim(-20, 120)
#plt.xlabel('Time (sec)', fontsize = 24)
#plt.ylabel('Temperature change rate \n(%, 100% at 0sec)', fontsize = 24)
plt.tick_params(labelbottom=False,
                labelleft=False,
                labelright=False,
                labeltop=False)



#温度緩和の時定数の計算
def relaxation(time, ratio, t1, t2):
    return ratio*np.exp(-time/t1) + (1-ratio)*np.exp(-time/t2)
result_100msec = curve_fit(relaxation, time_normal_100msec, average_normal_100msec[50:100])[0]
result_500msec = curve_fit(relaxation, time_normal_500msec, average_normal_500msec[50:100])[0]
result_1sec = curve_fit(relaxation, time_normal_1sec, average_normal_1sec[50:100])[0]
result_5sec = curve_fit(relaxation, time_normal_5sec, average_normal[100:150])[0]
result_10sec = curve_fit(relaxation, time_normal_10sec, average_normal_10sec[200:250])[0]

plt.xlim(-0.2, 3.5)
plt.ylim(-35, 110)

#fitting結果の図示
plt.plot(time_normal_100msec, relaxation(np.ravel(time_normal_100msec), result_100msec[0], result_100msec[1], result_100msec[2])*100, color = 'black')
plt.plot(time_normal_500msec, relaxation(np.ravel(time_normal_500msec), result_500msec[0], result_500msec[1], result_500msec[2])*100, color = 'blue')
plt.plot(time_normal_1sec, relaxation(np.ravel(time_normal_1sec), result_1sec[0], result_1sec[1], result_1sec[2])*100, color = 'green')
plt.plot(time_normal_5sec, relaxation(np.ravel(time_normal_5sec), result_5sec[0], result_5sec[1], result_5sec[2])*100, color = 'orange')
plt.plot(time_normal_10sec, relaxation(np.ravel(time_normal_10sec), result_10sec[0], result_10sec[1], result_10sec[2])*100, color = 'red')

#pdfのセーブ
pp = PdfPages('./heating_time_relaxation.pdf')
pp.savefig(fig)
pp.close()

#データの保存
pd.DataFrame(average_normal_100msec[50:100]*100/average_normal_100msec[50]).to_csv('100msec_normed.csv')
pd.DataFrame(average_normal_500msec[50:100]*100/average_normal_500msec[50]).to_csv('500msec_normed.csv')
pd.DataFrame(average_normal_1sec[50:100]*100/average_normal_1sec[50]).to_csv('1sec_normed.csv')
pd.DataFrame(average_normal[100:150]*100/average_normal[100]).to_csv('5sec_normed.csv')
pd.DataFrame(average_normal_10sec[200:250]*100/average_normal_10sec[200]).to_csv('10sec_normed.csv')

(sem_100msec)[50:100].to_csv('sem_100msec.csv')
(sem_500msec)[50:100].to_csv('sem_500msec.csv')
(sem_1sec)[50:100].to_csv('sem_1sec.csv')
(sem_5sec)[100:150].to_csv('sem_5sec.csv')
(sem_10sec)[200:250].to_csv('sem_10sec.csv')
```
