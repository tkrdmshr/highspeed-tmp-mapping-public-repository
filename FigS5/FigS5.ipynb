```python
import glob
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.backends.backend_pdf import PdfPages
```

```python
data_before = pd.read_csv('./250919_CP_heating/decay_curve/cell01_cyto_before.csv', sep = '\t')
before_param = pd.read_csv('./250919_CP_heating/decay_curve/fitting/01_C_b.csv')
data_during = pd.read_csv('./250919_CP_heating/decay_curve/cell01_cyto_during.csv', sep = '\t')
during_param = pd.read_csv('./250919_CP_heating/decay_curve/fitting/01_C_d.csv')
time = np.float32(data_before.iloc[1:509, 0])
decay_before = np.float32(data_before.iloc[1:509, 1])
decay_during = np.float32(data_during.iloc[1:509, 1])
```

```python
fig = plt.figure(figsize=(20, 20))

plt.rcParams['font.family'] = 'Arial'
plt.tick_params(which = 'minor', bottom = 'False',  direction = "out", labelsize = 40, length = 20)
plt.tick_params(bottom = 'False', axis='both', direction = 'out', labelsize = 40, length = 20)
plt.xlim(0, 50)
plt.ylim(1, 10**4)
plt.yscale('log')

#fitting lineの図示
t0 = before_param.loc[1]['Lifetime Offset ns']
Bkgr = before_param.loc[1]['Tail Offset Cnts']
tau1 = before_param.loc[1]['Lifetime (τ) 1 ns']
tau2 = before_param.loc[1]['Lifetime (τ) 2 ns']
A1 = before_param.loc[1]['Amplitude 1 kCnts']*1000
A2 = before_param.loc[1]['Amplitude 2 kCnts']*1000
print('短い蛍光寿命は%fns' % tau1)
print('長い蛍光寿命は%fns' % tau2)
print('短い蛍光寿命の割合は%f' % (A1/(A1+A2)))
print('長い蛍光寿命の割合は%f' % (A2/(A1+A2)))


def decay_curve(time):
    return A1*np.exp(-(time-t0)/tau1) + A2*np.exp(-(time-t0)/tau2) + Bkgr


plt.plot(time, decay_before,  marker = 'o' ,markerfacecolor="None", markeredgecolor='black', markersize = 10, markeredgewidth = 2, linewidth = 0)
plt.plot(time[26:494], decay_curve(time[26:494]), color = 'black', linewidth = 5)


#pdfのセーブ
pp = PdfPages('./CP_decay_curve_cyto_b.pdf')
pp.savefig(fig)
pp.close()
plt.show()

#データの保存
pd.DataFrame([time, decay_before]).to_csv('before_cyto_plot.csv')
```
