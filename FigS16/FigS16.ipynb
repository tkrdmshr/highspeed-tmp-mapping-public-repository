```python
import glob
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scipy as sp
from scipy.optimize import curve_fit
from matplotlib.backends.backend_pdf import PdfPages
```

```python
def figmake():
    fig = plt.figure(figsize = (20, 20))
    plt.rcParams['font.family'] = 'Arial'
    plt.rcParams['xtick.major.width'] = 2.0
    plt.rcParams['ytick.major.width'] = 2.0
    plt.rcParams['axes.linewidth'] = 2.0
    plt.tick_params(axis='both', direction = "in", labelsize = 24)
    return fig
```

```python
path = './02.csv'
intensity = pd.read_csv(path, header = None)[1][1:]
time = [0.07*(i-150) for i in range(len(intensity))]

fig = figmake()
plt.plot(time, intensity, color = 'black')
plt.xlim(-12, 12)
plt.ylim(0, 150)

#pdfのセーブ
pp = PdfPages('./FigS16_timelapse.pdf')
pp.savefig(fig)
pp.close()
```
