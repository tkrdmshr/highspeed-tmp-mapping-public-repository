```python
import pandas as pd
import glob
import numpy as np
import matplotlib.pyplot as plt
from calibration import lifetime_to_temperature_201112
from scipy.optimize import curve_fit
from matplotlib.backends.backend_pdf import PdfPages
```

```python
def figmake():
    plt.rcParams['font.family'] = 'Arial'
    plt.rcParams['xtick.major.width'] = 1.0
    plt.rcParams['ytick.major.width'] = 1.0
    plt.rcParams['axes.linewidth'] = 1.0
    plt.rcParams["font.size"] = 20
    plt.figure(figsize = (6, 6))
```

```python
file_path = glob.glob('./FCCP_relaxation/220*_FCCP_100uM_26_relaxation_130mW/FCCP/*min.csv')
file_path.sort()
date = ['0123_FCCP', '0123_2_FCCP', '0124_FCCP', '0124_2_FCCP', '0125_FCCP', '0126_FCCP', '0130_FCCP', '0131_FCCP', '0201_FCCP']
```

```python
#FCCPデータのdfへのまとめなおし
all_df = pd.DataFrame()
for day in date:
    result_df = pd.DataFrame()
    paths = [path for path in file_path if day in path]
    for path in paths:
        df = pd.read_csv(path,sep = '\t')
        tmp_df = df.iloc[1, 1::2]
        result_df = pd.concat([result_df, tmp_df.T], axis = 1)
        cell_number = int(result_df.shape[0]/2)
    all_df = pd.concat([all_df, result_df.T.iloc[:, :cell_number]], axis = 1)
```

```python
max_delta = []
FCCP_delta = []
figmake()
time = [0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60]
for num in range(int(all_df.shape[1])):

    tmp = lifetime_to_temperature_201112(np.array(all_df.iloc[:, num], dtype = 'float64'))
    delta = tmp - tmp[0]
    FCCP_delta.append(delta)
    max_delta.append(delta[-1])
    plt.xlim(-10, 70)
    plt.plot(time, delta, linewidth = 0, marker = 'o', markersize = 10, alpha = 0.35)

pd.DataFrame(FCCP_delta).to_csv('FCCP_delta.csv')
```

```python
file_path = glob.glob('./FCCP_relaxation/220*_FCCP_100uM_26_relaxation_130mW/relaxation/cell*.csv')
file_path.sort()
```

```python
half_time = []
count = 0
for day in date:
    paths = [path for path in file_path if day in path]
    cell_num = ['cell' + str(num).zfill(2) for num in range(1, int(len(paths)/5+1))]
    
    for num in cell_num:
        #print(count)
        count += 1
        cell_path = [path for path in paths if num in path]
        relaxation_ave = 0
        
        for path in cell_path:
            df = pd.read_csv(path, sep = '\t')
            relaxation_int = np.array(df.iloc[1:, 3], dtype = 'float64')
            relaxation = np.array(df.iloc[1:, 1], dtype = 'float64')
            diff = relaxation_int[:-1] - relaxation_int[1:]
            off_time = np.argmax(diff)
            relaxation = lifetime_to_temperature_201112(relaxation)
            relaxation_delta = relaxation - np.mean(relaxation[:50])
            relaxation_ave += relaxation_delta[off_time: off_time + 135]
            
        relaxation_ave = relaxation_ave/len(cell_path)
        relaxation_ave = relaxation_ave/relaxation_ave[0]
        time = [i*0.07 for i in range(len(relaxation_ave))]
        def relaxation(time, ratio, t1, t2):
            return ratio*np.exp(-time/t1) + (1-ratio)*np.exp(-time/t2)
        result = curve_fit(relaxation, np.array(time), relaxation_ave)[0]
        half_time.append(sum(relaxation_ave[2:50])*100*0.07)
        print(sum(relaxation_ave[2:50])*100*0.07)
        figmake()
        fig = plt.figure(figsize=(10, 10))
        plt.plot(time[:50], relaxation_ave[:50]*100, color = 'blue', linewidth = 0, marker = 'o', markersize = 10, alpha = 1)
        plt.plot(time, relaxation(np.array(time), result[0], result[1], result[2])*100)
        plt.xlim(-0.2, 3.5)
        plt.ylim(-35, 110)
        
        plt.show()
```

```python
df = pd.DataFrame([half_time, max_delta])
df.index = ['AUC', '∆T']
df.to_csv('AUC_vs_∆T.csv')
```
